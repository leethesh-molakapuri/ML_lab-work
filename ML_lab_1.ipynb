```python
# Import required libraries
from google.colab import files
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# 1. Upload Image
# --------------------------------------------------

uploaded = files.upload()

# Get the uploaded image name
file_name = list(uploaded.keys())[0]

# Open the image
img = Image.open(file_name).convert("RGB")

# Convert image into NumPy array
arr = np.array(img)

print("Original Image Shape:", arr.shape)
print("Number of Dimensions:", arr.ndim)
print("Minimum Pixel Value:", arr.min())
print("Maximum Pixel Value:", arr.max())
print("Average Brightness:", arr.mean())

# --------------------------------------------------
# 2. Display Original Image
# --------------------------------------------------

plt.figure(figsize=(5, 5))
plt.imshow(arr)
plt.title("Original Image")
plt.axis("off")
plt.show()


# --------------------------------------------------
# 3. Image Cropping
# --------------------------------------------------

# Crop using NumPy slicing
# rows: 100 to 400
# columns: 150 to 500
cropped = arr[100:400, 150:500]

plt.figure(figsize=(5, 5))
plt.imshow(cropped)
plt.title("Cropped Image")
plt.axis("off")
plt.show()

print("Cropped Image Shape:", cropped.shape)


# --------------------------------------------------
# 4. Image Flipping
# --------------------------------------------------

# Flip image horizontally
horizontal_flip = np.fliplr(arr)

# Flip image vertically
vertical_flip = np.flipud(arr)

plt.figure(figsize=(5, 5))
plt.imshow(horizontal_flip)
plt.title("Horizontally Flipped Image")
plt.axis("off")
plt.show()

plt.figure(figsize=(5, 5))
plt.imshow(vertical_flip)
plt.title("Vertically Flipped Image")
plt.axis("off")
plt.show()


# --------------------------------------------------
# 5. RGB Channel Analysis
# --------------------------------------------------

# Extract Red, Green and Blue channels
red = arr[:, :, 0]
green = arr[:, :, 1]
blue = arr[:, :, 2]

# Calculate average value of each channel
print("Average Red Value:", red.mean())
print("Average Green Value:", green.mean())
print("Average Blue Value:", blue.mean())

# Display RGB channels
plt.figure(figsize=(5, 5))
plt.imshow(red, cmap="gray")
plt.title("Red Channel")
plt.axis("off")
plt.show()

plt.figure(figsize=(5, 5))
plt.imshow(green, cmap="gray")
plt.title("Green Channel")
plt.axis("off")
plt.show()

plt.figure(figsize=(5, 5))
plt.imshow(blue, cmap="gray")
plt.title("Blue Channel")
plt.axis("off")
plt.show()


# --------------------------------------------------
# 6. Brightness Analysis
# --------------------------------------------------

brightness = arr.mean()

print("Average Image Brightness:", brightness)

if brightness < 85:
    print("Image is Dark")
elif brightness < 170:
    print("Image has Normal Brightness")
else:
    print("Image is Bright")


# --------------------------------------------------
# 7. Thresholding
# --------------------------------------------------

# Convert RGB image to grayscale
gray = np.mean(arr, axis=2)

# Set threshold value
threshold = 128

# Convert image into black and white
binary = gray > threshold

# Display thresholded image
plt.figure(figsize=(5, 5))
plt.imshow(binary, cmap="gray")
plt.title("Thresholded Image")
plt.axis("off")
plt.show()
```

```python
# PART-15: Step 1 - Text to Tokens                       Roll no : AP24110011803

# Store the given sentence
sentence2 = "Data Science and Machine Learning use NumPy arrays every single day"

# Convert the sentence to lowercase and split it into individual words
tokens2 = sentence2.lower().split()

# Print the tokens
print(tokens2)
```

```python
# Convert tokens2 into a NumPy array
tokens2_array = np.array(tokens2)

# Find all unique words
vocabulary2 = np.unique(tokens2_array)

# Print the vocabulary
print("Vocabulary:", vocabulary2)

# Print the number of unique words
print("Vocabulary size:", len(vocabulary2))
```

```python
# Create a function to build a Bag-of-Words vector
def bag_of_words_vector(tokens, vocabulary):
    # Count how many times each vocabulary word appears in the tokens
    vector = np.array([np.sum(np.array(tokens) == word) for word in vocabulary])

    # Return the word-frequency vector
    return vector
```

```python
# Create a bag-of-words vector for tokens2
vec2 = bag_of_words_vector(tokens2, vocabulary2)

# Print the vocabulary
print(vocabulary2)

# Print the word-frequency vector
print(vec2)
```

```python
# Similar pair of sentences
sentence1 = "I love learning Python programming"
sentence2 = "I enjoy learning Python programming"

# Convert sentences into tokens
tokens1 = sentence1.lower().split()
tokens2 = sentence2.lower().split()

# Create a common vocabulary
vocab1 = np.unique(np.array(tokens1 + tokens2))

# Create Bag-of-Words vectors
vec1 = bag_of_words_vector(tokens1, vocab1)
vec2 = bag_of_words_vector(tokens2, vocab1)

# Calculate cosine similarity
similarity1 = np.dot(vec1, vec2) / (
    np.linalg.norm(vec1) * np.linalg.norm(vec2)
)

print("Similar Pair Cosine Similarity:", similarity1)


# Very different pair of sentences
sentence3 = "I love playing football"
sentence4 = "The weather is cold today"

# Convert sentences into tokens
tokens3 = sentence3.lower().split()
tokens4 = sentence4.lower().split()

# Create a common vocabulary
vocab2 = np.unique(np.array(tokens3 + tokens4))

# Create Bag-of-Words vectors
vec3 = bag_of_words_vector(tokens3, vocab2)
vec4 = bag_of_words_vector(tokens4, vocab2)

# Calculate cosine similarity
similarity2 = np.dot(vec3, vec4) / (
    np.linalg.norm(vec3) * np.linalg.norm(vec4)
)

print("Different Pair Cosine Similarity:", similarity2)
```

```python
# Find the 3 Most Frequent Words

# Create a paragraph
paragraph = """
Python is a powerful programming language.
Python is easy to learn and Python is widely used.
Learning Python helps students build useful projects.
"""

# Tokenize the paragraph
tokens = paragraph.lower().split()

# Remove punctuation from each word
tokens = [word.strip(".,!?") for word in tokens]

# Build the vocabulary
vocabulary = np.unique(np.array(tokens))

# Build the word-count vector
word_counts = bag_of_words_vector(tokens, vocabulary)

# Find indices of the 3 most frequent words
top_indices = np.argsort(word_counts)[::-1][:3]

# Print the top 3 most frequent words
print("Top 3 Most Frequent Words:")

for i in top_indices:
    print(vocabulary[i], ":", word_counts[i])
```

```python
# Mini Project: Simple Text Similarity Checker

# Three sentences
sentence1 = "I love learning Python programming"
sentence2 = "I enjoy learning Python programming"
sentence3 = "The weather is beautiful today"

# Tokenize all three sentences
tokens1 = sentence1.lower().split()
tokens2 = sentence2.lower().split()
tokens3 = sentence3.lower().split()

# Create one shared vocabulary for all three sentences
shared_vocab = np.unique(np.array(tokens1 + tokens2 + tokens3))

# Convert each sentence into a Bag-of-Words vector
vec1 = bag_of_words_vector(tokens1, shared_vocab)
vec2 = bag_of_words_vector(tokens2, shared_vocab)
vec3 = bag_of_words_vector(tokens3, shared_vocab)

# Calculate cosine similarity between two vectors
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

# Calculate similarity for all three pairs
similarity_12 = cosine_similarity(vec1, vec2)
similarity_13 = cosine_similarity(vec1, vec3)
similarity_23 = cosine_similarity(vec2, vec3)

# Print the similarity scores
print("Similarity between Sentence 1 and Sentence 2:", similarity_12)
print("Similarity between Sentence 1 and Sentence 3:", similarity_13)
print("Similarity between Sentence 2 and Sentence 3:", similarity_23)

# Find the most similar pair
scores = {
    "Sentence 1 and Sentence 2": similarity_12,
    "Sentence 1 and Sentence 3": similarity_13,
    "Sentence 2 and Sentence 3": similarity_23
}

most_similar_pair = max(scores, key=scores.get)

print("\nMost Similar Pair:", most_similar_pair)
print("Highest Similarity Score:", scores[most_similar_pair])
```
